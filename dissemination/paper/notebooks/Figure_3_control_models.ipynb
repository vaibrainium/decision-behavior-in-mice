# Figure 3 — control-trial model comparison

Run **Setup** and **Shared data**, then run any labeled panel independently. Panels B–C are held out; panel D is descriptive and in-sample. Canonical code: `scripts/paper/fig3_control_models.py`.

## Setup

```python
%matplotlib inline
%load_ext autoreload
%autoreload 2
import matplotlib.pyplot as plt
from scripts.paper import fig3_control_models as F
from scripts.paper.style import apply_style, label_panel
apply_style()
```

## Shared data

```python
strategy, components = F.score_tables()
quality = F.load_or_simulate_quality()
strategy[['mouse', 'rig', 'delta_nll']]
```

## Panel A — candidate pipelines

```python
fig, ax = plt.subplots(figsize=(7.0, 3.6), layout='constrained')
F.panel_schematic(ax)
label_panel(ax, 'A', x=-0.07, y=1.04)
plt.show()
```

## Panel B — held-out strategy comparison

```python
fig, ax = plt.subplots(figsize=(6.0, 4.5), layout='constrained')
F.panel_strategy(ax, strategy)
label_panel(ax, 'B', x=-0.18, y=1.04)
plt.show()
```

## Panel C — required components and extra dynamics

```python
fig, axes = plt.subplots(1, 2, figsize=(9.0, 4.0), layout='constrained')
F.component_axis(axes[0], components, ['Remove sv', 'Remove fast guesses'], 'Required components')
F.component_axis(axes[1], components, ['Add leak', 'Add urgency', 'Add leak + urgency'], 'Extra dynamics')
axes[0].set_ylabel('Held-out NLL: comparator − strict model')
axes[1].set_ylabel('Held-out NLL: extension − strict model')
axes[0].set_ylim(-30, 540); axes[1].set_ylim(-50, 22)
label_panel(axes[0], 'C', x=-0.16, y=1.04)
plt.show()
```

## Panel D — final-fit quality on control trials

```python
fig, axes = plt.subplots(1, 3, figsize=(11.0, 3.8), layout='constrained')
F.panel_fit_quality(axes, quality)
axes[0].text(-0.10, 1.16, 'Final-fit quality on control trials (descriptive; in-sample)', transform=axes[0].transAxes, ha='left', va='bottom')
label_panel(axes[0], 'D', x=-0.18, y=1.12)
plt.show()
```
