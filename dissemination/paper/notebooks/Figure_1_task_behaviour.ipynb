## Setup

```python
%matplotlib inline
%load_ext autoreload
%autoreload 2
import matplotlib.pyplot as plt
from scripts.paper import fig1_task_behaviour as F
from scripts.paper.style import apply_style, label_panel
apply_style()
```

## Shared data

```python
data = F.load_trials()
ctrl = data[data['sched_pulse_type'] == 0].copy()
schedule = F.make_schedule(data)
psych_mouse, psych_group = F.psychometric_table(ctrl)
rt_mouse, rt_group = F.rt_table(ctrl)
nr_mouse, nr_group = F.no_response_table(ctrl)
F.overview_table(data, ctrl)
```

## Panel A — task schematic

```python
fig, ax = plt.subplots(figsize=(7.0, 4.2), layout='constrained')
F.panel_task(ax)
label_panel(ax, 'A', x=-0.08, y=1.04)
plt.show()
```

## Panel B — scheduled pulse timing

```python
fig, ax = plt.subplots(figsize=(8.0, 4.0), layout='constrained')
F.panel_schedule(ax, schedule)
label_panel(ax, 'B', x=-0.08, y=1.04)
plt.show()
```

## Panel C — psychometric function

```python
fig, ax = plt.subplots(figsize=(5.5, 4.2), layout='constrained')
F.panel_psychometric(ax, psych_mouse, psych_group)
label_panel(ax, 'C', x=-0.16, y=1.04)
plt.show()
```

## Panel D — response time by outcome

```python
fig, ax = plt.subplots(figsize=(5.5, 4.2), layout='constrained')
F.panel_rt(ax, rt_mouse, rt_group)
label_panel(ax, 'D', x=-0.16, y=1.04)
plt.show()
```

## Panel E — no response by 5 s

```python
fig, ax = plt.subplots(figsize=(5.5, 4.2), layout='constrained')
F.panel_no_response(ax, nr_mouse, nr_group)
label_panel(ax, 'E', x=-0.16, y=1.04)
plt.show()
```
